```python
primes = [2, 3, 5, 7]
for p in primes:
    square = p ** 2
    cubed = p ** 3
    tetrated = p ** 4
    millioned = p ** 6
    billioned = p ** 9
    trillioned = p ** 12
    quadrillioned = p ** 15
    googled = p ** 100
    googledplexed = p ** 1000
    print(square, cubed, tetrated, millioned, billioned, trillioned, quadrillioned, googled, googledplexed)
```
